```python
movies = [
    {"name": "Avengers", "genre": "Action", "rating": 8.0},
    {"name": "Iron Man", "genre": "Action", "rating": 7.9},
    {"name": "Titanic", "genre": "Romance", "rating": 7.8},
    {"name": "The Notebook", "genre": "Romance", "rating": 7.9},
    {"name": "Inception", "genre": "Sci-Fi", "rating": 8.8},
    {"name": "Interstellar", "genre": "Sci-Fi", "rating": 8.6},
    {"name": "The Conjuring", "genre": "Horror", "rating": 7.5},
    {"name": "Annabelle", "genre": "Horror", "rating": 5.8}
]


def search_by_genre():
    genre = input("Enter genre: ").lower()

    print("\nMovies in", genre.title(), "genre:")

    found = False

    for movie in movies:
        if movie["genre"].lower() == genre:
            print(movie["name"], "- Rating:", movie["rating"])
            found = True

    if not found:
        print("No movies found.")


def recommend_movies():
    movie_name = input("Enter movie name: ").lower()

    selected_movie = None

    for movie in movies:
        if movie["name"].lower() == movie_name:
            selected_movie = movie
            break

    if selected_movie is None:
        print("Movie not found.")
        return

    print("\nRecommended Movies:")

    for movie in movies:
        if (movie["genre"] == selected_movie["genre"] and
                movie["name"].lower() != movie_name):

            print(movie["name"], "- Rating:", movie["rating"])


while True:

    print("\n===== MOVIE RECOMMENDATION SYSTEM =====")
    print("1. Display Movies")
    print("2. Search by Genre")
    print("3. Recommend Similar Movies")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("\nAvailable Movies:")

        for movie in movies:
            print(
                movie["name"],
                "| Genre:", movie["genre"],
                "| Rating:", movie["rating"]
            )

    elif choice == "2":
        search_by_genre()

    elif choice == "3":
        recommend_movies()

    elif choice == "4":
        print("Thank you!")
        break

    else:
        print("Invalid choice!")
```
